### Character slection

To be able to filter the different characters and update dynamically the plots in this section, we have used an artificial heatmap that allows clicking on the desired characters. This is based on the following implementation: https://altair-viz.github.io/gallery/interactive_column_selection.html, but for 1-dimensional selections. This heatmap uses character faces to avoid requiring users to read labels. Default values are: Homer, Marge, Bart, Lisa, and Mr Burns.

Limitations: We encountered some issues when rendering the chart in Streamlit with these interactions, which caused the plot to blink constantly. To solve this, we converted the chart to HTML, which made the plot height dependent on the initial character selection. This means that selecting more characters does not increase the plot size, and therefore some characters will be hidden. We were unable to implement a limit on selections and instead specified a maximum of 5 in the title. Despite these limitations, we believe the trade-off is beneficial for us an the users: visual selection through character faces is more user-friendly than a basic Streamlit multiselect, and also follows the goal of the course of learning Altair. That is why we decided to keep this filter approach.

### Q3 - I would like to compare the word distribution for a pair of concrete characters, for a selected season, and, in addition, for a selected episode.

To compare the distribution of dialogue between two characters across episodes and seasons, we utilize an interactive lollipop chart. This visualization enables a direct, side-by-side comparison, making differences in character contributions easily identifiable.

The interface features two character selectors on the left side, allowing users to choose the exact pair they wish to compare. Additionally, a horizontally aligned season selector at the top filters the data to focus on a specific season.

The visualization offers two distinct comparison modes: absolute and relative. In absolute mode, the x-axis displays the total words spoken, while the y-axis represents the episode number. Each episode features a data point accompanied by the character's icon, allowing users to track the absolute dialogue volume for each character throughout the season.

In relative mode, the chart highlights the exact difference in word counts. The x-axis is centered at zero, which indicates a tie. Values extending to either side reveal which character spoke more and by what margin. The icon displayed at the end of each lollipop identifies the dominant character for that specific episode.

Together, these modes allow users to analyze both raw dialogue output and relative conversational dominance. One current limitation is that users can select the same character twice, and the UI permits multiple selections, potentially causing ambiguous comparisons.

### Q4 - I would like the same comparison as before, but over a concrete episode.


To answer this question, we interpret it as understanding how the dialogue is distributed within a specific episode and how the selected characters contribute throughout its progression.

We use a cumulative line chart to visualize this information. The x-axis represents the line number within the episode, which indicates the progression of the episode over time. The y-axis represents the cumulative number of words spoken by each character.

This visualization allows us to compare two selected characters throughout the episode. The slope of each line indicates how fast a character accumulates dialogue at a given moment, while the distance between the lines highlights differences in speaking activity. The final value of each line represents the total number of words spoken by that character during the episode, enabling a direct comparison of their overall participation.

The chart is linked to the episode comparison view on the left. By hovering over a specific episode, the cumulative line chart updates to display the dialogue progression for that episode only. This interaction allows users to examine how dialogue is distributed within individual episodes and directly answers the question of how word counts evolve over the course of a particular episode.
