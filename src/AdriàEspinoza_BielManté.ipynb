## Section 1

### Character slection

To be able to filter the different characters and update dynamically the plots in this section, we have used an artificial heatmap that allows clicking on the desired characters. This is based on the following implementation: https://altair-viz.github.io/gallery/interactive_column_selection.html, but for 1-dimensional selections. This heatmap uses character faces to avoid requiring users to read labels. Default values are: Homer, Marge, Bart, Lisa, and Mr Burns.

Limitations: We encountered some issues when rendering the chart in Streamlit with these interactions, which caused the plot to blink constantly. To solve this, we converted the chart to HTML, which made the plot height dependent on the initial character selection. This means that selecting more characters does not increase the plot size, and therefore some characters will be hidden. We were unable to implement a limit on selections and instead specified a maximum of 5 in the title. Despite these limitations, we believe the trade-off is beneficial for us an the users: visual selection through character faces is more user-friendly than a basic Streamlit multiselect, and also follows the goal of the course of learning Altair. That is why we decided to keep this filter approach.

### Q1 - What are the characters that issue more words, and how are the word numbers distributed?

To solve the first part of the question we have chosen a barplot, ordered by word frequency, and for the second part we have chosen a jitter plot, where each point represents the number of words per epsiode. These 2 plots have been aligned horizontally so it's easy to connect the total word counts of a charcater to it's dsitribution across epsiodes withot the need of nameing the y axis twice. Additionally we have used chatrcater faces instead of labels on the y-axis, since it allows the user to not have to read a whole label, and make a quicker connection to which character the each row represents.

### Q2 -  How have the word numbers among the characters evolved throughout the available seasons?

The x axis represents the individual seasons, and the y axis represents the aggregation of words per character for each of them. Each line color represnenst a character, specified in the legend. It is also possible to filter for more detail on the selected characters by interacting with the legend.
